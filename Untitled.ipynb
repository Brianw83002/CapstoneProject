```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from sklearn.model_selection import train_test_split
from torch.utils.data import Subset
```

```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=[.5, .5, .5], std=[.5, .5, .5])
])

#change path to where ever you have data, make sure it points to parent folder containing the subfolders of split data
dataset = datasets.ImageFolder(root=r'C:\Users\maxpd\Desktop\RDD_Data\21431547\RDD2022_released_through_CRDDC2022\RDD2022\United_States\train\exports_cls_zip',
                               transform=transform
                              )
#verification to know if dataset did actually load
print(f"Classes found: {dataset.classes}")
```

```python
#splitting data into train, test, validate. 70/15/15
targets = dataset.targets
indices = list(range(len(dataset)))

train_val_idx, test_idx = train_test_split(
    indices,
    test_size=.15,
    stratify=targets,
    random_state=42
)
train_val_targets = [targets[i] for i in train_val_idx]

train_idx, val_idx = train_test_split(
    train_val_idx,
    test_size=.176,
    stratify=train_val_targets,
    random_state=42
)
train_set = Subset(dataset, train_idx)
val_set = Subset(dataset, val_idx)
test_set = Subset(dataset, test_idx)
#verification
print(f"Train: {len(train_set)}, Val: {len(val_set)}, Test: {len(test_set)}")
```

```python
#data loaders
train_loader = DataLoader(train_set, batch_size=32, shuffle=True)
val_loader = DataLoader(val_set, batch_size=32, shuffle=False)
test_loader = DataLoader(test_set, batch_size=32, shuffle=False)
```

```python
#creating CNN
class RoadDetectionCNN(nn.Module):
    def __init__(self, num_classes=4):
        super(RoadDetectionCNN, self).__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            
            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
        )
        
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256 * 8 * 8, 512),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )
    
    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

#enables GPU training
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
model = RoadDetectionCNN(num_classes=4).to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=.001)
```

```python
from tqdm import tqdm

num_epochs = 25
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0

    loop = tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}')
    for images, labels in loop:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        loop.set_postfix(loss=loss.item())

    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    print(f'Epoch {epoch+1}/{num_epochs}, Loss: {running_loss/len(train_loader):.4f}, Val Acc: {100*correct/total:.2f}%')
```

```python
torch.save(model.state_dict(), 'road_classification_model.pth')
print("model saved")
```

```python
model.eval()
correct, total = 0, 0
with torch.no_grad():
    for images, label in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        total += label.size(0)
        correct += (predicted == label).sum().item()

print(f'Accuracy: {100*correct/total:.2f}%')
```
